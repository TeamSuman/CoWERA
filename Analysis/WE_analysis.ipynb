### Is the data converged ? If "yes", calculate MFPT by running the following cell.

```python
# MFPT via the Hill relation (Eq. 2 of the manuscript).
#
# BIAS FIX -- read before quoting a number.
# The running curve below is evaluated AT EACH WARP, so its time axis is built from the
# warp cycle indices. The FINAL estimate must NOT be read off the last point of that
# curve: simulated time AFTER the last warp was paid for and produced no flux, so
# dropping it from the denominator inflates the rate. Measured on real runs: up to a
# 1.6x inflation (one replica's last warp fell at cycle 3082 of 5000 -> 38% of its
# sampling silently discarded). Always divide by the TOTAL simulated time.

def _n_cycles_run(h5path):
    """Actual number of WE cycles completed -- NOT the requested n_cycles, and NOT the
    cycle index of the last warp."""
    import h5py
    with h5py.File(h5path, 'r') as f:
        if 'runs/0/progress/distances_fromtarget' in f:
            return int(len(f['runs/0/progress/distances_fromtarget']))
        if 'runs/0/resampling/_cycle_idxs' in f:
            return int(np.max(f['runs/0/resampling/_cycle_idxs'][()])) + 1
    raise RuntimeError('could not determine the number of cycles run')

n_cycles_run = _n_cycles_run(filename)
T_total = n_cycles_run * delta_t * 0.001 * n_walkers     # aggregate simulated time (ns)

mfpt_running = (T_list / np.cumsum(w_i)) * 0.001         # running estimate (us)
mfpt_final   = (T_total / np.sum(w_i))   * 0.001         # <-- UNBIASED final estimate (us)
rate_final_1e7 = (1.0 / (mfpt_final * 1e-6)) / 1e7

plt.plot(T_list / n_walkers, mfpt_running, lw=1.5, marker='o', color='#fdb462', markersize=3)
plt.axhline(mfpt_final, ls='--', lw=1, color='k')
plt.xlabel("Cumulative simulation time (\u03bcs)")
plt.ylabel("MFPT (\u03bcs)")
plt.text(0.5, 0.8, f'MFPT = {mfpt_final:.3f} \u03bcs', transform=plt.gca().transAxes)
plt.yscale("log")

dead = 1.0 - cycle_idxs[-1] / n_cycles_run
print(f"cycles run          : {n_cycles_run}")
print(f"last warp at cycle  : {int(cycle_idxs[-1])}  ({100*dead:.1f}% of the run produced no further flux)")
print(f"MFPT (total time)   : {mfpt_final:.4f} us    <-- report this")
print(f"MFPT (last-warp)    : {mfpt_running[-1]:.4f} us    (legacy convention; MFPT biased low, rate biased high)")
print(f"rate (total time)   : {rate_final_1e7:.4f} x1e7 s^-1")
```

### Calculate rate by running the following cell

```python
import numpy as np
import matplotlib.pyplot as plt
import os
from wepy.hdf5 import WepyHDF5

print(f"Processing: {filename}")

wepy_h5 = WepyHDF5(filename, mode='r')
wepy_h5.open()

# ---- Infer timestep from folder name (optional) ----
folder_name = os.path.basename(os.path.dirname(filename))
if "steps" in folder_name:
    steps_str = folder_name.split("steps")[1].split("_")[0]
    delta_t = int(steps_str) * 0.002  # ps
else:
    delta_t = 5000 * 0.002  # fallback (ps)

# ---- Extract warping data ----
n_walkers = wepy_h5.num_init_walkers(0)
cycle_idxs = np.array(wepy_h5.h5['/runs/0/warping/_cycle_idxs'])
weights = np.array([x[0] for x in wepy_h5.h5['/runs/0/warping/weight']])
wepy_h5.close()

def _n_cycles_run(h5path):
    """Actual number of WE cycles completed -- NOT the requested n_cycles, and NOT the
    cycle index of the last warp."""
    import h5py
    with h5py.File(h5path, 'r') as f:
        if 'runs/0/progress/distances_fromtarget' in f:
            return int(len(f['runs/0/progress/distances_fromtarget']))
        if 'runs/0/resampling/_cycle_idxs' in f:
            return int(np.max(f['runs/0/resampling/_cycle_idxs'][()])) + 1
    raise RuntimeError('could not determine the number of cycles run')

# ---- Molecular time (ns) at each warp, for the RUNNING curve ----
T_list = cycle_idxs * delta_t * 1e-3

# ---- Rate: running curve + UNBIASED final estimate ----
# The final value must use the TOTAL simulated time, not the time of the last warp
# (see the bias note in the MFPT cell above).
n_cycles_run = _n_cycles_run(filename)
mfpt_running = (T_list * n_walkers / np.cumsum(weights)) * 1e-9        # s
rate_running_1e7 = (1.0 / mfpt_running) / 1e7

mfpt_final = (n_cycles_run * delta_t * 1e-3 * n_walkers / np.sum(weights)) * 1e-9
rate_final_1e7 = (1.0 / mfpt_final) / 1e7

fig, ax = plt.subplots()
ax.plot(T_list, rate_running_1e7, lw=2.5, marker='o', label="CoWERA (running)")
ax.axhline(rate_final_1e7, ls='--', lw=1.5, color='k',
           label=f"final = {rate_final_1e7:.3f}" + r"$\times 10^7\ s^{-1}$")
ax.set_xlabel(r'$\mathrm{Molecular\ Time\ (ns)}$')
ax.set_ylabel(r'$\mathrm{Rate\ (10^7\ s^{-1})}$')
ax.set_yscale("log")
ax.grid(True)
ax.legend()
plt.tight_layout()
plt.show()

print(f"rate (total time) : {rate_final_1e7:.4f} x1e7 s^-1   <-- report this")
print(f"rate (last-warp)  : {rate_running_1e7[-1]:.4f} x1e7 s^-1   (legacy, biased high)")
```
